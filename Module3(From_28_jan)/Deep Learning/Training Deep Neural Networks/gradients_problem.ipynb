### <div style='color: #068120'>***Here are some of the problems you could run into when you train a deep neural network 🔴***</div>

- *You may be faced with the problem of gradients growing ever smaller or larger when flowing backward through the DNN during training. Both of these problems make lower layers very hard to train.*

- *You might not have enough training data for such a large network, or it might be too costly to label.*

- *Training may be extremely slow.*

- *A model with millions of parameters risks severely overfitting the training set, especially if there are not enough training instances or if they are too noisy.*

---
1. `Vanishing Gradient Problem` - *the gradient descent update leaves the lower layers’ connection weights virtually unchanged, and training never converges to a good solution.*

<div style='line-height: 1.6; text-align: center'> 
During training, a neural network calculates its error and sends it backward from the output layer to the input layer using the chain rule of calculus. This involves multiplying multiple derivatives together. If these derivative values are consistently small (e.g., less than), multiplying them repeatedly layer after layer causes the overall gradient to shrink exponentially.
</div>

**<div style='color: #05967e; font-size: 30px'>Key culprits include:</div>**

1. *<div style='color: #ff0404'>Sigmoid & Tanh Activation Functions:</div> These functions compress data into a specific range. Their maximum derivative is only 0.25, meaning repeated multiplication quickly shrinks the gradient toward zero.*

2. *<div style='color: #ff0404'>Network Depth:</div> The more layers a network has, the more derivatives are multiplied, making deeper layers highly susceptible to vanishing gradients.*

3. *<div style='color: #ff0404'>Recurrent Neural Networks (RNNs):</div>When processing long sequences, data must be passed backward through many time steps. This acts like an extremely deep network, causing early historical context to be "forgotten" or ignored.*

---
2. `Exploding  Gradient Problem` - *a challenge in deep learning where gradients accumulate during backpropagation and become excessively large.*

<div style='line-height: 1.6; text-align: center'> 
During training, a neural network calculates the gradient using the chain rule, multiplying derivatives from each layer back to the beginning. If these derivative values consistently average greater than 1.0, the repeated multiplication acts like an exponential snowball effect. This is especially common in architectures with many layers (like deep CNNs) or long-term sequence dependencies (like vanilla RNNs).
</div>
